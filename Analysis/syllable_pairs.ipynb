```python
coda_identity
```

```python
phonetic_features_list = [onset_place, onset_manner,  vowel_place, vowel_manner, vowel_identity, coda_place, coda_manner,  onset_cluster, coda_cluster,onset_identity, coda_identity]
```

```python
phonetic_titles = ["Onset Place", "Onset Manner", "Vowel Place", "Vowel Manner", "Vowel Identity", "Coda Place", "Coda Manner", "Onset Cluster", "Coda Cluster","Onset Identity", "Coda Identity"]
```

```python
import matplotlib.pyplot as plt
import dissimilarity as ds


for i in range(len(phonetic_features_list)):

    plt.figure(figsize=(18, 18))

    plt.title(f"Distance Matrix for {phonetic_titles[i]} Features", fontsize=20)

    ax = sns.heatmap(ds.distances_matrix(phonetic_features_list[i]), xticklabels=unique_stimuli , yticklabels=unique_stimuli, cmap="YlGnBu", annot=False, fmt=".2f")
```

```python
model_matrix = [ds.distances_matrix(phonetic_features_list[i]) for i in range(len(phonetic_features_list))]

plt.figure(figsize=(18, 18))
plt.title("Distance Matrix for Summed Phonetic Features", fontsize=20)
ax = sns.heatmap(np.sum(model_matrix, axis=0), xticklabels=unique_stimuli , yticklabels=unique_stimuli, cmap="YlGnBu", annot=False, fmt=".2f")
```

```python
binarized_matrix = model_matrix.copy()
```

```python
for i in range(len(model_matrix)):
    for j in range(len(model_matrix[i])):
        for k in range(len(model_matrix[i][j])):
            if model_matrix[i][j][k] > 0:
                binarized_matrix[i][j][k] = 1

display(binarized_matrix)
```

```python
plt.figure(figsize=(18, 18))
plt.title("Binarized Matrix for Summed Phonetic Features", fontsize=20)
ax = sns.heatmap(np.sum(binarized_matrix, axis=0), xticklabels=unique_stimuli , yticklabels=unique_stimuli, cmap="YlGnBu", annot=False, fmt=".2f")
```

```python
binarized_sum = np.sum(binarized_matrix, axis=0)

display(binarized_sum)
```

```python
promising_coordinates = []

for i in range(len(binarized_sum)):
    for j in range(len(binarized_sum[i])):
        if binarized_sum[i][j] == 10:
            promising_coordinates.append((i,j))

promising_coordinates
```

```python
promising_pairs = []

for i in range(len(stimuli["syll_IPA"])):
    for j in range(len(stimuli["syll_IPA"])):
        if (i,j) in promising_coordinates:
            promising_pairs.append((stimuli.iloc[i]['syll_IPA'], stimuli.iloc[j]['syll_IPA']))

promising_pairs
```

```python
import random

promising_samples = []

for i in range(20):
    promising_samples.append(random.sample(promising_pairs, k=50))
```

```python
import dissimilarity as ds


plt.figure(figsize=[22,22])
plt.title("Distances Between Syllables Under Each Model")
ax = sns.heatmap(ds.make_predictions(promising_pairs[:50]), yticklabels=promising_pairs[:50],xticklabels=["P_N_N","M_N_N","I_N_N","N_P_N","N_M_N","N_I_N","N_N_P","N_N_M","N_N_I","C_N_N","N_N_C"])
```

```python


for i in range(len(promising_samples)):
    plt.figure(figsize=(16, 16))

    plt.title("Correlation Coefficients of Pairs of Syllables Under All Models")

    ax = sns.heatmap(ds.experiment_metric(ds.make_predictions(promising_samples[i])),xticklabels=["P_N_N","M_N_N","I_N_N","N_P_N","N_M_N","N_I_N","N_N_P","N_N_M","N_N_I","C_N_N","N_N_C"],yticklabels=["P_N_N","M_N_N","I_N_N","N_P_N","N_M_N","N_I_N","N_N_P","N_N_M","N_N_I","C_N_N","N_N_C"])
    print(promising_samples[i])
```
