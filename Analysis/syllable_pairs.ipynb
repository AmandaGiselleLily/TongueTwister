```python
def vector_distance(vector1, vector2):
    return np.linalg.norm(np.array(vector1) - np.array(vector2))
```

```python
def distances_matrix(vectors):
    n = len(vectors)
    distance_matrix = np.zeros((n, n))
    
    for i in range(n):
        for j in range(n):
            distance_matrix[i][j] = vector_distance(vectors[i], vectors[j])
    
    return distance_matrix
```

```python
phonetic_features_list = [onset_place, onset_manner,  vowel_place, vowel_manner, vowel_identity, coda_place, coda_manner,  onset_cluster, coda_cluster,onset_identity, coda_identity]
```

```python
phonetic_titles = ["Onset Place", "Onset Manner", "Vowel Place", "Vowel Manner", "Vowel Identity", "Coda Place", "Coda Manner", "Onset Cluster", "Coda Cluster","Onset Identity", "Coda Identity"]
```

```python
import matplotlib.pyplot as plt


for i in range(len(phonetic_features_list)):

    plt.figure(figsize=(18, 18))

    plt.title(f"Distance Matrix for {phonetic_titles[i]} Features", fontsize=20)

    ax = sns.heatmap(distances_matrix(phonetic_features_list[i]), xticklabels=unique_stimuli , yticklabels=unique_stimuli, cmap="YlGnBu", annot=False, fmt=".2f")
```

```python
model_matrix = [distances_matrix(phonetic_features_list[i]) for i in range(len(phonetic_features_list))]

plt.figure(figsize=(18, 18))
plt.title("Distance Matrix for Summed Phonetic Features", fontsize=20)
ax = sns.heatmap(np.sum(model_matrix, axis=0), xticklabels=unique_stimuli , yticklabels=unique_stimuli, cmap="YlGnBu", annot=False, fmt=".2f")
```
