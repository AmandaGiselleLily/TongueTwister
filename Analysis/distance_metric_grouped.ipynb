```python
def cluster_consonant(string1,string2):
    
    vowels = "əʊaiɪeɛuʊɔæɒᵊ"
    nucleus1 = -1
    nucleus2 = -1
    clusters1 = {"onset": False, "coda": False}
    clusters2 = {"onset": False, "coda": False}

    for char in vowels:
        if string1.find(char) != -1:
            nucleus1 = char
            break
    for char in vowels:
        if string2.find(char) != -1:
            nucleus2 = char
            break

    construct1 = string1.split(nucleus1)
    construct2 = string2.split(nucleus2)

    if len(construct1[0]) > 1:
        clusters1["onset"] = True
    if len(construct1[1]) > 1:
        clusters1["coda"] = True

    if len(construct2[0]) > 1:
        clusters2["onset"] = True
    if len(construct2[1]) > 1:
        clusters2["coda"] = True

    return np.linalg.norm(np.array(list(clusters1.values()), dtype=int) - np.array(list(clusters2.values()), dtype=int))


```

```python
def articulation_manner(string1,string2):
    manner_space1 = {
        'plosive': [0,0,0,0,0,0],
        'nasal': [0,0,0,0,0,0],
        'fricative': [0,0,0,0,0,0],
        'affricate': [0,0,0,0,0,0],
        'approximant': [0,0,0,0,0,0]
    }

    manner_space2 = {
            'plosive': [0,0,0,0,0,0],
            'nasal': [0,0,0,0,0,0],
            'fricative': [0,0,0,0,0,0],
            'affricate': [0,0,0,0,0,0],
            'approximant': [0,0,0,0,0,0]
        }

    plosives = "pbtdkg"
    nasals = "mnŋ"
    fricatives = "fvθðh"
    affricates = "ʤʧ"
    approximants = "wjlɹ"

    for i in range(len(string1)):
        char = string1[i]
        if char in plosives:
            manner_space1["plosive"][i] = 1
        if char in nasals:
            manner_space1["nasal"][i] = 1
        if char in fricatives:
            manner_space1["fricative"][i] = 1
        if char in affricates:
            manner_space1["affricate"][i] = 1
        if char in approximants:
            manner_space1["approximant"][i] = 1

    for j in range(len(string2)):
        char = string2[j]
        if char in plosives:
            manner_space2["plosive"][j] = 1
        if char in nasals:
            manner_space2["nasal"][j] = 1
        if char in fricatives:
            manner_space2["fricative"][j] = 1
        if char in affricates:
            manner_space2["affricate"][j] = 1
        if char in approximants:
            manner_space2["approximant"][j] = 1

    return np.linalg.norm(np.array(list(manner_space1.values())).flatten() - np.array(list(manner_space2.values())).flatten())
```

```python
def articulation_place(string1,string2):

    articulation_space1 = {'bilabial': [0,0,0,0,0,0],
                           'labiodental': [0,0,0,0,0,0],
                           'dental': [0,0,0,0,0,0],
                           'alveolar': [0,0,0,0,0,0],
                           'palatal': [0,0,0,0,0,0],
                           'velar': [0,0,0,0,0,0],
                           'uvular': [0,0,0,0,0,0],
                           'pharyngeal': [0,0,0,0,0,0],
                           'glottal': [0,0,0,0,0,0]}
    
    articulation_space2 = {'bilabial': [0,0,0,0,0,0],
                           'labiodental': [0,0,0,0,0,0],
                           'dental': [0,0,0,0,0,0],
                           'alveolar': [0,0,0,0,0,0],
                           'palatal': [0,0,0,0,0,0],
                           'velar': [0,0,0,0,0,0],
                           'uvular': [0,0,0,0,0,0],
                           'pharyngeal': [0,0,0,0,0,0],
                           'glottal': [0,0,0,0,0,0]}

    bilabial = "mp"
    labiodental = "bf"
    dental = "td"
    alveolar = "sz"
    palatal = "tʃdʒ"
    velar = "kɡ"
    uvular = "qɢ"
    pharyngeal = "ħʕ"
    glottal = "hʔ"


    for i in range(len(string1)):
        char = string1[i]
        if char in bilabial:
            articulation_space1["bilabial"][i] = 1
        if char in labiodental:
            articulation_space1["labiodental"][i] = 1
        if char in dental:
            articulation_space1["dental"][i] = 1
        if char in alveolar:
            articulation_space1["alveolar"][i] = 1
        if char in palatal:
            articulation_space1["palatal"][i] = 1
        if char in velar:
            articulation_space1["velar"][i] = 1
        if char in uvular:
            articulation_space1["uvular"][i] = 1
        if char in pharyngeal:
            articulation_space1["pharyngeal"][i] = 1
        if char in glottal:
            articulation_space1["glottal"][i] = 1
    for j in range(len(string2)):  
        char = string2[j]
        if char in bilabial:
            articulation_space2["bilabial"][j] = 1
        if char in labiodental:
            articulation_space2["labiodental"][j] = 1
        if char in dental:
            articulation_space2["dental"][j] = 1
        if char in alveolar:
            articulation_space2["alveolar"][j] = 1
        if char in palatal:
            articulation_space2["palatal"][j] = 1
        if char in velar:
            articulation_space2["velar"][j] = 1
        if char in uvular:
            articulation_space2["uvular"][j] = 1
        if char in pharyngeal:
            articulation_space2["pharyngeal"][j] = 1
        if char in glottal:
            articulation_space2["glottal"][j] = 1


    return np.linalg.norm(np.array(list(articulation_space1.values())).flatten() - np.array(list(articulation_space2.values())).flatten())
    
    
```
