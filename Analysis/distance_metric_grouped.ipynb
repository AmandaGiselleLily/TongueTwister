```python
from pickletools import string1

import numpy as np

import Levenshtein

def phonemic_distance(string1, string2):

    nasals = "mnŋ"
    fricatives = "fvθðh"
    plosives = "pbtdkg"
    sibilants = "sʃzʒʤʧ"
    rhotics = "r"
    vowels = "aəʊiɪeɛuʊɔæɒᵊ"
    laterals = "l"

    syl_space1 = {"n1":0,"f1":0,"p1":0,"s1":0,"r1":0,"v1":0,"l1":0,"n2":0,"f2":0,"p2":0,"s2":0,"r2":0,"v2":0,"l2":0,"n3":0,"f3":0,"p3":0,"s3":0,"r3":0,"v3":0,"l3":0,"n4":0,"f4":0,"p4":0,"s4":0,"r4":0,"v4":0,"l4":0,"n5":0,"f5":0,"p5":0,"s5":0,"r5":0,"v5":0,"l5":0,"n6":0,"f6":0,"p6":0,"s6":0,"r6":0,"v6":0, "l6":0}
    syl_space2 = {"n1":0,"f1":0,"p1":0,"s1":0,"r1":0,"v1":0,"l1":0,"n2":0,"f2":0,"p2":0,"s2":0,"r2":0,"v2":0,"l2":0,"n3":0,"f3":0,"p3":0,"s3":0,"r3":0,"v3":0,"l3":0,"n4":0,"f4":0,"p4":0,"s4":0,"r4":0,"v4":0,"l4":0,"n5":0,"f5":0,"p5":0,"s5":0,"r5":0,"v5":0,"l5":0,"n6":0,"f6":0,"p6":0,"s6":0,"r6":0,"v6":0, "l6":0}

    for char in string1:

        if string1.index(char) == 0:
            if char in nasals:
                syl_space1["n1"] += 1
            if char in fricatives:
                syl_space1["f1"] += 1
            if char in plosives:
                syl_space1["p1"] += 1
            if char in sibilants:
                syl_space1["s1"] += 1
            if char in rhotics:
                syl_space1["r1"] += 1
            if char in vowels:
                syl_space1["v1"] += 1
            if char in laterals:
                syl_space1["l1"] += 1

        if string1.index(char) == 1:
        
            if char in nasals:
                syl_space1["n2"] += 1
            if char in fricatives:
                syl_space1["f2"] += 1
            if char in plosives:
                syl_space1["p2"] += 1
            if char in sibilants:
                syl_space1["s2"] += 1
            if char in rhotics:
                syl_space1["r2"] += 1
            if char in vowels:
                syl_space1["v2"] += 1
            if char in laterals:
                syl_space1["l2"] += 1

        if string1.index(char) == 2:

            if char in nasals:
                syl_space1["n3"] += 1
            if char in fricatives:
                syl_space1["f3"] += 1
            if char in plosives:
                syl_space1["p3"] += 1
            if char in sibilants:
                syl_space1["s3"] += 1
            if char in rhotics:
                syl_space1["r3"] += 1
            if char in vowels:
                syl_space1["v3"] += 1
            if char in laterals:
                syl_space1["l3"] += 1

        if string1.index(char) == 3:

            if char in nasals:
                    syl_space1["n4"] += 1
            if char in fricatives:
                    syl_space1["f4"] += 1
            if char in plosives:
                    syl_space1["p4"] += 1
            if char in sibilants:
                    syl_space1["s4"] += 1
            if char in rhotics:
                    syl_space1["r4"] += 1
            if char in vowels:
                    syl_space1["v4"] += 1
            if char in laterals:
                    syl_space1["l4"] += 1

        if string1.index(char) == 4:

            if char in nasals:
                syl_space1["n5"] += 1
            if char in fricatives:
                syl_space1["f5"] += 1
            if char in plosives:
                syl_space1["p5"] += 1
            if char in sibilants:
                syl_space1["s5"] += 1
            if char in rhotics:
                syl_space1["r5"] += 1
            if char in vowels:
                syl_space1["v5"] += 1
            if char in laterals:
                syl_space1["l5"] += 1

        if string1.index(char) == 5:

            if char in nasals:
                syl_space1["n6"] += 1
            if char in fricatives:
                syl_space1["f6"] += 1
            if char in plosives:
                syl_space1["p6"] += 1
            if char in sibilants:
                syl_space1["s6"] += 1
            if char in rhotics:
                syl_space1["r6"] += 1
            if char in vowels:
                syl_space1["v6"] += 1
            if char in laterals:
                syl_space1["l6"] += 1

    for char in string2:
        if string2.index(char) == 0:
            if char in nasals:
                syl_space2["n1"] += 1
            if char in fricatives:
                syl_space2["f1"] += 1
            if char in plosives:
                syl_space2["p1"] += 1
            if char in sibilants:
                syl_space2["s1"] += 1
            if char in rhotics:
                syl_space2["r1"] += 1
            if char in vowels:
                syl_space2["v1"] += 1
            if char in laterals:
                syl_space2["l1"] += 1

        if string2.index(char) == 1:
        
            if char in nasals:
                syl_space2["n2"] += 1
            if char in fricatives:
                syl_space2["f2"] += 1
            if char in plosives:
                syl_space2["p2"] += 1
            if char in sibilants:
                syl_space2["s2"] += 1
            if char in rhotics:
                syl_space2["r2"] += 1
            if char in vowels:
                syl_space2["v2"] += 1
            if char in laterals:
                syl_space2["l2"] += 1

        if string2.index(char) == 2:

            if char in nasals:
                syl_space2["n3"] += 1
            if char in fricatives:
                syl_space2["f3"] += 1
            if char in plosives:
                syl_space2["p3"] += 1
            if char in sibilants:
                syl_space2["s3"] += 1
            if char in rhotics:
                syl_space2["r3"] += 1
            if char in vowels:
                syl_space2["v3"] += 1
            if char in laterals:
                syl_space2["l3"] += 1

        if string2.index(char) == 3:

            if char in nasals:
                    syl_space2["n4"] += 1
            if char in fricatives:
                    syl_space2["f4"] += 1
            if char in plosives:
                    syl_space2["p4"] += 1
            if char in sibilants:
                    syl_space2["s4"] += 1
            if char in rhotics:
                    syl_space2["r4"] += 1
            if char in vowels:
                    syl_space2["v4"] += 1
            if char in laterals:
                    syl_space2["l4"] += 1

        if string2.index(char) == 4:

            if char in nasals:
                syl_space2["n5"] += 1
            if char in fricatives:
                syl_space2["f5"] += 1
            if char in plosives:
                syl_space2["p5"] += 1
            if char in sibilants:
                syl_space2["s5"] += 1
            if char in rhotics:
                syl_space2["r5"] += 1
            if char in vowels:
                syl_space2["v5"] += 1
            if char in laterals:
                syl_space2["l5"] += 1

        if string2.index(char) == 5:

            if char in nasals:
                syl_space2["n6"] += 1
            if char in fricatives:
                syl_space2["f6"] += 1
            if char in plosives:
                syl_space2["p6"] += 1
            if char in sibilants:
                syl_space2["s6"] += 1
            if char in rhotics:
                syl_space2["r6"] += 1
            if char in vowels:
                syl_space2["v6"] += 1
            if char in laterals:
                syl_space2["l6"] += 1

    #print(syl_space1)
    #print(syl_space2)
    return float(np.linalg.norm(np.array(list(syl_space1.values())) - np.array(list(syl_space2.values()))))


```
